```python
import pandas as pd
from pathlib import Path

file_path = Path("../data/raw/census/dp05.csv")

df_dp05 = pd.read_csv(file_path, dtype=str)

print("Dataset shape:", df_dp05.shape)
print("\nFirst 10 rows:")
display(df_dp05.head(10))

print("\nColumn names:")
print(df_dp05.columns.tolist())
```

```python
print("Dataset shape:", df_dp05.shape)

print("\nData types:")
print(df_dp05.dtypes.value_counts())

print("\nExact duplicate rows:", df_dp05.duplicated().sum())

print("\nUnique labels:", df_dp05["Label (Grouping)"].nunique())

print("\nMissing values per column:")
print(df_dp05.isna().sum())

print("\nFirst 30 demographic labels:")
for i, label in enumerate(df_dp05["Label (Grouping)"].head(30)):
    print(i, ":", repr(label))
```

```python
labels_dp05 = df_dp05["Label (Grouping)"].str.strip()

for i, label in enumerate(labels_dp05):
    print(i, ":", repr(label))
```

```python
duplicate_rows = df_dp05[df_dp05.duplicated(keep=False)]

display(duplicate_rows[["Label (Grouping)"]])
```
